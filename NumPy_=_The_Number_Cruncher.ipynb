```python
import numpy as np

scores = np.random.randint(0, 100, size=50)

print("Scores:")
print(scores)

print("\nMean:", np.mean(scores))
print("Median:", np.median(scores))
print("Highest:", np.max(scores))
print("Lowest:", np.min(scores))
print("Standard Deviation:", np.std(scores))

print("\n25th Percentile:", np.percentile(scores, 25))
print("50th Percentile:", np.percentile(scores, 50))
print("75th Percentile:", np.percentile(scores, 75))

failed = scores[scores < 40]
distinction = scores[scores > 85]

print("\nFailed Students:")
print(failed)

print("\nDistinction Students:")
print(distinction)

normalized = (scores - np.min(scores)) / (np.max(scores) - np.min(scores))

print("\nNormalized Scores:")
print(normalized)

reshaped = scores.reshape(5, 10)

print("\nReshaped Array:")
print(reshaped)

subject_avg = np.mean(reshaped, axis=1)

print("\nSubject-wise Averages:")
print(subject_avg)

theory = np.random.randint(0, 100, size=50)
practical = np.random.randint(0, 100, size=50)

final_scores = (0.4 * theory) + (0.6 * practical)

print("\nWeighted Final Scores:")
print(final_scores)
```
